## Import necessary Libraries

```python
import string
import ipywidgets as widgets
from IPython.display import display, clear_output

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

## Faq Dataset giving

```python
faqs = {
    "What is artificial intelligence?":
        "Artificial Intelligence is the ability of machines to simulate human intelligence.",
    "What is machine learning?":
        "Machine learning is a subset of AI that allows systems to learn from data.",
    "What is deep learning?":
        "Deep learning is a type of machine learning using neural networks.",
    "What is NLP?":
        "Natural Language Processing enables computers to understand human language.",
    "What is Python used for?":
        "Python is used in AI, data science, web development, and automation."
}
```

## NLP Preprocessing

```python
stop_words = set(stopwords.words('english'))

def preprocess(text):
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words]
    return " ".join(tokens)
```

## TF-IDF Vectorization

```python
questions = list(faqs.keys())
processed_questions = [preprocess(q) for q in questions]

vectorizer = TfidfVectorizer()
faq_vectors = vectorizer.fit_transform(processed_questions)
```

## Chatbot Response Function

```python
def chatbot_response(user_input):
    user_input = preprocess(user_input)
    user_vector = vectorizer.transform([user_input])
    similarity_scores = cosine_similarity(user_vector, faq_vectors)
    best_match_index = similarity_scores.argmax()
    return faqs[questions[best_match_index]]
```

## Simple GUI

```python
# GUI widgets
input_box = widgets.Text(
    placeholder='Ask a question...',
    description='You:',
    layout=widgets.Layout(width='70%')
)

ask_button = widgets.Button(description='Ask')
output_area = widgets.Output()

def on_button_clicked(b):
    with output_area:
        clear_output()
        user_input = input_box.value
        if user_input.strip() == "":
            print("Please enter a question.")
        else:
            response = chatbot_response(user_input)
            print("Chatbot:", response)

ask_button.on_click(on_button_clicked)

display(input_box, ask_button, output_area)
```
