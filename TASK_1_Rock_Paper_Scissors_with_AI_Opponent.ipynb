### Task 1 Rock-Paper-Scissors with AI Opponent

```python
import random

options = ["rock", "paper", "scissors"]

print("*** Rock Paper Scissors Game with AI Opponent ***")
player = input("Enter your Preferred choice (rock/paper/scissors): ").lower()

if player not in options:
    print("Invalid choice! Try again.")
else:
    ai = random.choice(options)
    print(f"AI chose: {ai}")

    # Determine the result
    if player == ai:
        print("It's a draw!")
    elif (player == "rock" and ai == "scissors") or \
         (player == "paper" and ai == "rock") or \
         (player == "scissors" and ai == "paper"):
        print("You win!")
    else:
        print("You lose!")
```

### Task 1 Bonus Section

```python
import random

options = ["rock", "paper", "scissors"]

# Track user move frequencies (AI memory)
user_history = {
    "rock": 0,
    "paper": 0,
    "scissors": 0
}

print("*** Rock Paper Scissors Game with AI Opponent (Learning Mode) ***")

while True:
    player = input("\nEnter your choice (rock/paper/scissors or quit): ").lower()

    if player == "quit":
        print("Game ended. Thanks for playing!")
        break

    if player not in options:
        print("Invalid choice! Try again.")
        continue


    user_history[player] += 1


    if sum(user_history.values()) > 1:

        predicted_move = max(user_history, key=user_history.get)

        if predicted_move == "rock":
            ai = "paper"
        elif predicted_move == "paper":
            ai = "scissors"
        else:
            ai = "rock"
    else:

        ai = random.choice(options)

    print(f"AI chose: {ai}")


    if player == ai:
        print("It's a draw!")
    elif (player == "rock" and ai == "scissors") or \
         (player == "paper" and ai == "rock") or \
         (player == "scissors" and ai == "paper"):
        print("You win!")
    else:
        print("You lose!")

    print("AI Learning Stats:", user_history)
```
